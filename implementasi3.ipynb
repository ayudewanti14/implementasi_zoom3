```python

data = {
    "ID": list(range(1, 16)),
    "Nama_Aset": [
        "Jalan A", "Jalan B", "Jalan C",
        "Jembatan A", "Jembatan B",
        "Drainase A", "Drainase B", "Drainase C",
        "Taman A", "Taman B", "Taman C",
        "Lampu Jalan A", "Lampu Jalan B",
        "Halte A", "Halte B"
    ],
    "Jenis_Aset": [
        "Jalan", "Jalan", "Jalan",
        "Jembatan", "Jembatan",
        "Drainase", "Drainase", "Drainase",
        "Taman", "Taman", "Taman",
        "Lampu Jalan", "Lampu Jalan",
        "Halte", "Halte"
    ],
    "Usia_Tahun": [
        15, 8, 20, 25, 10,
        12, 18, 5, 7, 14,
        3, 9, 16, 11, 6
    ],
    "Kondisi": [
        35, 75, 25, 30, 80,
        45, 20, 90, 85, 40,
        95, 60, 30, 55, 88
    ],
    "Penggunaan": [
        90, 70, 95, 85, 60,
        80, 90, 50, 65, 75,
        40, 85, 95, 70, 45
    ]
}

df = pd.DataFrame(data)

print("DATA ASET INFRASTRUKTUR")
display(df)

df.to_csv("dataset_aset.csv", index=False)
```

```python

# Normalisasi nilai menjadi skala 0-100
df["Skor_Kondisi"] = 100 - df["Kondisi"]

usia_maksimum = 30
df["Skor_Usia"] = (
    df["Usia_Tahun"] / usia_maksimum * 100
).clip(0, 100)

df["Skor_Penggunaan"] = df["Penggunaan"]

# Menghitung weighted scoring
df["Skor_Prioritas"] = (
    0.50 * df["Skor_Kondisi"] +
    0.30 * df["Skor_Usia"] +
    0.20 * df["Skor_Penggunaan"]
).round(2)

display(
    df[[
        "Nama_Aset",
        "Skor_Kondisi",
        "Skor_Usia",
        "Skor_Penggunaan",
        "Skor_Prioritas"
    ]]
)
```

```python

def klasifikasi_prioritas(skor):
    if skor >= 70:
        return "Tinggi"
    elif skor >= 40:
        return "Sedang"
    else:
        return "Rendah"

df["Prioritas"] = df["Skor_Prioritas"].apply(
    klasifikasi_prioritas
)

df = df.sort_values(
    by="Skor_Prioritas",
    ascending=False
).reset_index(drop=True)

df["Peringkat"] = np.arange(1, len(df) + 1)

print("HASIL ANALISIS PRIORITAS PEMELIHARAAN")
display(
    df[[
        "Peringkat",
        "Nama_Aset",
        "Jenis_Aset",
        "Skor_Prioritas",
        "Prioritas"
    ]]
)
```

```python

warna = {
    "Tinggi": "#E74C3C",
    "Sedang": "#F39C12",
    "Rendah": "#2ECC71"
}

warna_bar = df["Prioritas"].map(warna)

plt.figure(figsize=(12, 7))

plt.barh(
    df["Nama_Aset"],
    df["Skor_Prioritas"],
    color=warna_bar
)

plt.gca().invert_yaxis()

plt.title(
    "Peringkat Prioritas Pemeliharaan Aset Infrastruktur",
    fontsize=14,
    fontweight="bold"
)
plt.xlabel("Skor Prioritas")
plt.ylabel("Nama Aset")
plt.xlim(0, 105)

plt.tight_layout()
plt.savefig(
    "grafik_peringkat_aset.png",
    dpi=200,
    bbox_inches="tight"
)
plt.show()
```

```python

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Usia_Tahun",
    y="Kondisi",
    hue="Prioritas",
    size="Penggunaan",
    hue_order=["Tinggi", "Sedang", "Rendah"],
    palette=warna,
    sizes=(50, 300),
    alpha=0.8
)

plt.title("Hubungan Usia, Kondisi, dan Prioritas Aset")
plt.xlabel("Usia Aset (Tahun)")
plt.ylabel("Kondisi Aset (0-100)")

plt.tight_layout()
plt.savefig("grafik_analisis_aset.png", dpi=200)
plt.show()
```

```python

def rekomendasi(prioritas):
    if prioritas == "Tinggi":
        return "Segera inspeksi dan rencanakan perbaikan"
    elif prioritas == "Sedang":
        return "Jadwalkan inspeksi dan pemeliharaan"
    else:
        return "Lakukan pemantauan rutin"

df["Rekomendasi"] = df["Prioritas"].apply(rekomendasi)

print("REKOMENDASI PEMELIHARAAN ASET")

display(
    df[[
        "Peringkat",
        "Nama_Aset",
        "Skor_Prioritas",
        "Prioritas",
        "Rekomendasi"
    ]]
)

df.to_csv(
    "hasil_analisis_prioritas_aset.csv",
    index=False
)
```
